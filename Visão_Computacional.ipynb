# Processamento de Imagens com OpenCV e Matplotlib

Este notebook apresenta um guia prático e sequencial contendo as principais técnicas de manipulação e processamento de imagem utilizando a biblioteca **OpenCV** e exibição com **Matplotlib**.

## 🛠️ Tecnologias Utilizadas
- **Python**
- **OpenCV (`opencv-python`)**
- **Matplotlib**

---

## 📋 Etapas do Projeto

1. **Configuração do Ambiente**: Instalação e importação das dependências necessárias.
2. **Carregamento e Exibição**: Leitura de imagem local convertendo o padrão de cores BGR (OpenCV) para RGB (Matplotlib).
3. **Conversão para Escala de Cinza**: Transformação da imagem colorida para Tons de Cinza (`grayscale`).
4. **Redimensionamento**: Alteração da resolução da imagem para dimensões específicas (`640x600`).
5. **Suavização (Blur)**: Aplicação do filtro Gaussiano para redução de ruídos e suavização dos detalhes.
6. **Detecção de Bordas**: Utilização do algoritmo de Canny para destacar contornos e linhas proeminentes.
7. **Desenho de Formas Geométricas**: Marcação e desenho de caixas delimitadoras (retângulos) diretamente na matriz da imagem.
8. **Salvamento do Resultado**: Exportação e gravação da imagem processada no armazenamento do ambiente.

```python
import cv2
import matplotlib.pyplot as plt

#Carregar a imagem
imagem = cv2.imread('/content/20260918_164351.jpg')

#Exibir imagem (apenas em IDE)
#cv2.imshow('Imagem', imagem)
#cv2.waitKey(0)
#cv2.destroyAllWindows()

#Converter a imagem de BGR para RGB
imagem_rgb = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)

#Exibir a imagem
plt.imshow(imagem_rgb)
plt.axis('off') #desabilita eixos
plt.show()
```

```python
# Converter a imagem colorida para grayscale.
img = "/content/20260918_164351.jpg"
```

```python
# Carregar a imagem
imagem = cv2.imread(img)

# Converter a imagem de BGR para RGB
imagem_gray = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

# Exibir a imagem cinza
plt.imshow(imagem_gray, cmap='gray')
plt.axis('off') #desabilita eixos
plt.show()
```

```python
#Redimensionamento de Imagem
## Alterar a resolução

# Redimensionar a imagem
largura = 640
altura = 600
dimensoes = (largura, altura)
imagem_redimensionada = cv2.resize(imagem, dimensoes, interpolation=cv2.INTER_AREA)

#Converter a imagem de BGR para RGB
imagem_rgb = cv2.cvtColor(imagem_redimensionada, cv2.COLOR_BGR2RGB)

# Exibir a imagem cinza
plt.imshow(imagem_rgb)
plt.axis('off') #desabilita eixos
plt.show()
```

```python
# Aplicação de filtros para suavização de imagens

#Aplicando suavização com filtro Gaussiano
imagem_suavizada = cv2.GaussianBlur(imagem_redimensionada, (15, 15), 0)

# Converter de BGR para RGB
imagem_rgb = cv2.cvtColor(imagem_suavizada, cv2.COLOR_BGR2RGB)

# Exibir a imagem cinza
plt.imshow(imagem_rgb)
plt.axis('off') #desabilita eixos
plt.show()
```

```python
# Detecção de bordas com Canny

#aplicando canny
imagem_bordas = cv2.Canny(imagem_redimensionada, 100, 200)

#convertendo de BGR para RGB
imagem_rgb = cv2.cvtColor(imagem_bordas, cv2.COLOR_BGR2RGB)

#exibindo a imagem
plt.imshow(imagem_rgb, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# Desenhar Formas Geometricas na Imagem

#Definir as coordenadas e a cor do retângulo
inicio = (2100, 1000) ## Deslocamento do inida esq -> dir, deslocamento de cima -> baixo
fim = (800, 2100) ## Deslocamento do fim esq -> dir, deslocamento de baixo -> cima
cor = (255, 0, 0) ## Azul em RGB
espessura = 10

#desenhar o retangulo na cópia da imagem
imagem_retangulo = cv2.rectangle(imagem.copy(), inicio, fim, cor, espessura)

#Converter a img BGR para RGB
imagem_rgb = cv2.cvtColor(imagem_retangulo, cv2.COLOR_BGR2RGB)

#Exibir a imagem
plt.imshow(imagem_rgb)
#plt.axis('off')
plt.show()
```

```python
# Salvando Imagem

salvar = '/content/foto1.jpg'
cv2.imwrite(salvar, imagem_bordas)
```
